### Самообъяснение ошибки порядка ячеек

**Почему ошибка появляется не из-за print, а из-за состояния runtime?**
Функция `print()` сама по себе исправна. Ошибка `NameError` возникае, так как в текущем состоянии среды выполнения (`runtime state`) имя `project_title` ещё не существует на момент запуска команды. Python выполняет команды последовательно: он не умеет заглядывать в ячейки, расположенные ниже на экране, пока пользователь физически их не запустит. Память компьютера на момент старта старой Ячейки 1 была пуста.

**Какой минимальный порядок ячеек исправляет ситуацию?**
Ситуацию исправляет перестановка ячеек местами: сначала обязательно должна выполняться инструкция присваивания значения переменной (`project_title = "Corpus Mini Lab"`), а только затем ячейка с функцией вывода `print(project_title)`.

---

| Выполняемая строка кода | Что меняется в памяти (Runtime State) | Визуальный вывод на экране |
| :--- | :--- | :--- |
| `project_title = "Corpus Mini Lab"` | В памяти создается строковая переменная с именем `project_title`. | Ничего не выводится. |
| `print(project_title)` | Состояние памяти не меняется. Происходит только чтение значения из переменной. | Выводится текст: `Corpus Mini Lab` |

## Часть 2. Средний уровень: Анализ лингвистической текстовой карточки

### 1. Обычный случай (Корректное выполнение)
Ниже создается переменная `text_card`, имитирующая лингвистическую карточку с термином. Мы выводим её содержимое и считаем количество символов. При последовательном запуске сверху вниз этот код всегда отрабатывает без ошибок.

```python
# Создаём переменную в runtime и сохраняем туда строку
text_card = "Лингвистический корпус текстов"
print("Запуск: обычный случай")
print(f"Содержимое карточки: {text_card}")
print(f"Длина строки: {len(text_card)} символов")
```

### 2. Ошибочный случай
Если сейчас принудительно перезапустить среду выполнения через меню (`Runtime -> Restart runtime`), то вся оперативная память очистится.

Если после перезапуска runtime сначала запустить ячейку, которая обращается к text_card, переменная ещё не существует и возникает NameError. Минимальное исправление: сначала нужно выполнить ячейку создания переменной, затем ячейку вывода.

```python
print(text_card)
```
